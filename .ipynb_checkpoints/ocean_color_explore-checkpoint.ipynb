# Tracking Ocean Life Based on its Color: exploratory graphs
Ocean color → **chlorophyll-a** (mg/m³), a proxy for phytoplankton, the base of the ocean food web.
We pull MODIS-Aqua chlorophyll and sea surface temperature (SST) from NASA GIBS, convert the colored
images back into values using each layer's colormap, and explore patterns off the California coast.

Only needs requests, numpy, pandas, matplotlib, pillow (all preinstalled in Colab). Runtime → Run all.

```python
import io, re, warnings
from concurrent.futures import ThreadPoolExecutor
from functools import lru_cache
import xml.etree.ElementTree as ET

import numpy as np
import pandas as pd
import requests
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from PIL import Image

warnings.filterwarnings("ignore", category=RuntimeWarning)  # nanmean on all-NaN slices

WMS  = "https://gibs.earthdata.nasa.gov/wms/epsg4326/best/wms.cgi"
CAPS = "https://gibs.earthdata.nasa.gov/wmts/epsg4326/best/1.0.0/WMTSCapabilities.xml"
```

## Config

```python
# Boxes are (lat_min, lon_min, lat_max, lon_max)
REGION = (30.0, -130.0, 42.0, -116.0)   # California Current: Baja → Oregon border
GLOBAL = (-90.0, -180.0, 90.0, 180.0)
RES = 0.05                              # degrees per pixel (~5 km)
YEARS = range(2019, 2024)
MAP_YEAR, MAP_MONTH = 2021, 5           # month used for the snapshot maps (May = spring upwelling)
NEARSHORE_KM = 50                       # "coastal" = within this distance of shore
OFFSHORE_KM = (200, 500)                # "open ocean" band
```

## Helpers: find layers, download images, decode colors → values

```python
_caps = None
def _capabilities():
    global _caps
    if _caps is None:
        _caps = requests.get(CAPS, timeout=180).text
    return _caps

def find_layers(*keywords):
    """List MODIS layer names containing all keywords, e.g. find_layers('Chlorophyll')."""
    ids = set(re.findall(r"<ows:Identifier>([^<]+)</ows:Identifier>", _capabilities()))
    return sorted(i for i in ids if "MODIS" in i and all(k.lower() in i.lower() for k in keywords))

def pick_layer(must, avoid=(), prefer=("Monthly", "8Day", "8_Day", "Daily")):
    """Choose the best-matching GIBS layer (prefers monthly/8-day composites: fewer cloud gaps)."""
    cands = find_layers(*must)
    if not cands:
        raise ValueError(f"No layer matches {must}")
    good = [c for c in cands if not any(a.lower() in c.lower() for a in avoid)] or cands
    rank = lambda c: next((i for i, p in enumerate(prefer) if p.lower() in c.lower()), len(prefer))
    return sorted(good, key=lambda c: (rank(c), len(c)))[0]

def colormap_url(layer):
    text = _capabilities()
    i = text.find(f"<ows:Identifier>{layer}</ows:Identifier>")
    if i >= 0:
        block = text[text.rfind("<Layer>", 0, i): text.find("</Layer>", i)]
        urls = re.findall(r'xlink:href="([^"]*colormaps[^"]*\.xml)"', block)
        if urls:
            return ([u for u in urls if "v1.3" in u] or urls)[0]
    return f"https://gibs.earthdata.nasa.gov/colormaps/v1.3/{layer}.xml"

_NUM = r"[-+]?INF|[-+]?(?:\d+\.?\d*|\.\d+)(?:[eE][-+]?\d+)?"
def _parse_value(s):
    """'[0.1,0.2)' -> midpoint, '[20,+INF)' -> 20, '5' -> 5."""
    if not s:
        return None
    nums = [float(n.replace("INF", "inf")) for n in re.findall(_NUM, s)]
    finite = [n for n in nums if np.isfinite(n)]
    if not finite:
        return None
    if len(finite) == 2 and finite[0] > 0 and finite[1] > 0:
        return float(np.sqrt(finite[0] * finite[1]))   # geometric midpoint (chlorophyll bins are log-spaced)
    return float(np.mean(finite))

def _tag(el):
    return el.tag.split("}")[-1]

@lru_cache(maxsize=None)
def load_colormap(layer):
    root = ET.fromstring(requests.get(colormap_url(layer), timeout=60).content)
    rgbs, vals, units = [], [], ""
    for cmap in (e for e in root.iter() if _tag(e) == "ColorMap"):
        units = cmap.get("units") or units
        for e in (x for x in cmap.iter() if _tag(x) == "ColorMapEntry"):
            if e.get("transparent") == "true" or e.get("nodata") == "true":
                continue
            v = _parse_value(e.get("value") or e.get("label"))
            if v is None or not e.get("rgb"):
                continue
            rgbs.append([int(c) for c in e.get("rgb").split(",")[:3]])
            vals.append(v)
    if not vals:
        raise ValueError(f"No usable colormap entries for {layer}")
    return np.array(rgbs, float), np.array(vals, float), units

def get_png(layer, date, bbox, res):
    lat0, lon0, lat1, lon1 = bbox
    params = dict(SERVICE="WMS", REQUEST="GetMap", VERSION="1.3.0", LAYERS=layer,
                  STYLES="", CRS="EPSG:4326", BBOX=f"{lat0},{lon0},{lat1},{lon1}",
                  WIDTH=int(round((lon1 - lon0) / res)), HEIGHT=int(round((lat1 - lat0) / res)),
                  FORMAT="image/png", TRANSPARENT="TRUE", TIME=date)
    r = requests.get(WMS, params=params, timeout=120)
    r.raise_for_status()
    if "image" not in r.headers.get("Content-Type", ""):
        raise RuntimeError(r.text[:300])
    return np.array(Image.open(io.BytesIO(r.content)).convert("RGBA"))

def decode(rgba, layer):
    """Nearest colormap color → data value for each pixel. Land / clouds (transparent) → NaN."""
    cm_rgb, cm_val, _ = load_colormap(layer)
    h, w, _ = rgba.shape
    px = rgba.reshape(-1, 4)
    out = np.full(len(px), np.nan)
    ok = px[:, 3] > 0
    rgb = px[ok, :3].astype(float)
    idx = np.empty(len(rgb), int)
    for s in range(0, len(rgb), 20000):
        d = ((rgb[s:s+20000, None, :] - cm_rgb[None]) ** 2).sum(-1)
        idx[s:s+20000] = d.argmin(1)
    out[ok] = cm_val[idx]
    return out.reshape(h, w)

@lru_cache(maxsize=None)
def values(layer, date, bbox=REGION, res=RES):
    try:
        return decode(get_png(layer, date, bbox, res), layer)
    except Exception as e:
        print(f"  skipped {layer} {date}: {e}")
        return None

def _days(layer):
    return (1,) if "monthly" in layer.lower() else (1, 8, 15, 22)

def monthly_grid(layer, year, month, bbox=REGION, res=RES):
    arrs = [values(layer, f"{year}-{month:02d}-{d:02d}", bbox, res) for d in _days(layer)]
    arrs = [a for a in arrs if a is not None]
    return np.nanmean(np.stack(arrs), axis=0) if arrs else None

def monthly_stack(layer, years=YEARS):
    """All months as a (time, lat, lon) array + matching dates."""
    jobs = [(y, m) for y in years for m in range(1, 13)]
    with ThreadPoolExecutor(8) as ex:
        grids = list(ex.map(lambda ym: monthly_grid(layer, *ym), jobs))
    shape = next(g.shape for g in grids if g is not None)
    stack = np.stack([g if g is not None else np.full(shape, np.nan) for g in grids])
    return stack, pd.to_datetime([f"{y}-{m:02d}-01" for y, m in jobs])

def to_celsius(a):
    return a - 273.15 if np.nanmedian(a) > 150 else a

def logmean(a, axis=None):
    """Chlorophyll is log-normally distributed, so average in log space (geometric mean)."""
    a = np.where(a > 0, a, np.nan)
    return 10 ** np.nanmean(np.log10(a), axis=axis)

def extent(bbox):
    return [bbox[1], bbox[3], bbox[0], bbox[2]]

def map_axes(ax, bbox):
    ax.set_facecolor("#d9d4c7")   # land / no data shows as tan
    ax.set_xlabel("Longitude"); ax.set_ylabel("Latitude")

CHL_NORM = LogNorm(0.03, 20)
```

## Pick layers and check colormaps
If anything fails here, run `find_layers("Chlorophyll")` / `find_layers("SST")` and set CHL / SST by hand.

```python
CHL = pick_layer(["Aqua", "Chlorophyll"], avoid=["L2"])
SST = pick_layer(["Aqua", "SST"], avoid=["L2", "Night", "MidIR"])
for L in (CHL, SST):
    rgb, val, units = load_colormap(L)
    print(f"{L}: {len(val)} colors, {val.min():.3g} to {val.max():.3g} {units}")
```

## Download monthly data for the region (takes a few minutes the first time)

```python
chl, dates = monthly_stack(CHL)
sst, _ = monthly_stack(SST)
sst = to_celsius(sst)

# Distance from shore: land is always NaN, so the easternmost ocean pixel in each row is the coastline
H, W = chl.shape[1:]
lats = REGION[2] - (np.arange(H) + 0.5) * RES
ocean = ~np.all(np.isnan(chl), axis=0)
coast_col = np.array([np.where(r)[0].max() if r.any() else -1 for r in ocean])
cols = np.arange(W)[None, :]
dist_km = (coast_col[:, None] - cols) * RES * 111.32 * np.cos(np.radians(lats))[:, None]
dist_km = np.where(ocean & (coast_col[:, None] >= 0) & (dist_km >= 0), dist_km, np.nan)
near = dist_km <= NEARSHORE_KM
off = (dist_km >= OFFSHORE_KM[0]) & (dist_km <= OFFSHORE_KM[1])
print(chl.shape, "| months with chlorophyll data:", int(np.sum(~np.all(np.isnan(chl), axis=(1, 2)))))
```

## Graph 1: Global ocean chlorophyll (where is ocean life concentrated?)

```python
g = monthly_grid(CHL, MAP_YEAR, MAP_MONTH, GLOBAL, 0.25)
fig, ax = plt.subplots(figsize=(13, 6.5))
im = ax.imshow(g, extent=extent(GLOBAL), cmap="viridis", norm=CHL_NORM)
map_axes(ax, GLOBAL)
lat0, lon0, lat1, lon1 = REGION
ax.add_patch(plt.Rectangle((lon0, lat0), lon1 - lon0, lat1 - lat0, fill=False, ec="red", lw=1.5))
fig.colorbar(im, ax=ax, label="Chlorophyll-a (mg/m³, log scale)", shrink=.8)
ax.set_title(f"Global ocean chlorophyll, {MAP_YEAR}-{MAP_MONTH:02d} (red box = our study region)")
plt.show()
```

## Graph 2: California coast. Chlorophyll vs. sea surface temperature

```python
i = list(dates).index(pd.Timestamp(MAP_YEAR, MAP_MONTH, 1))
fig, ax = plt.subplots(1, 2, figsize=(13, 6))
im0 = ax[0].imshow(chl[i], extent=extent(REGION), cmap="viridis", norm=CHL_NORM)
fig.colorbar(im0, ax=ax[0], label="Chlorophyll-a (mg/m³)")
ax[0].set_title(f"Chlorophyll, {MAP_YEAR}-{MAP_MONTH:02d}")
im1 = ax[1].imshow(sst[i], extent=extent(REGION), cmap="RdYlBu_r")
fig.colorbar(im1, ax=ax[1], label="°C")
ax[1].set_title(f"Sea surface temperature, {MAP_YEAR}-{MAP_MONTH:02d}")
for a in ax:
    map_axes(a, REGION)
plt.tight_layout(); plt.show()
```

## Graph 3: Coastal vs. open-ocean chlorophyll over time

```python
chl_near = np.array([logmean(m[near]) for m in chl])
chl_off = np.array([logmean(m[off]) for m in chl])
plt.figure(figsize=(12, 4.5))
plt.plot(dates, chl_near, marker="o", ms=3, color="seagreen", label=f"Coastal (< {NEARSHORE_KM} km)")
plt.plot(dates, chl_off, marker="o", ms=3, color="steelblue", label=f"Open ocean ({OFFSHORE_KM[0]}-{OFFSHORE_KM[1]} km)")
plt.yscale("log"); plt.ylabel("Chlorophyll-a (mg/m³, geometric mean)")
plt.title("Monthly chlorophyll: coast vs. open ocean"); plt.legend(); plt.grid(alpha=.3, which="both")
plt.show()
```

## Graph 4: Seasonal cycle of coastal chlorophyll, one line per year

```python
plt.figure(figsize=(9, 4.5))
for y in YEARS:
    sel = dates.year == y
    plt.plot(dates[sel].month, chl_near[sel], marker="o", ms=3, label=y)
plt.xticks(range(1, 13), ["J","F","M","A","M","J","J","A","S","O","N","D"])
plt.yscale("log"); plt.ylabel("Chlorophyll-a (mg/m³)")
plt.title("Coastal chlorophyll by month (spring/summer upwelling?)"); plt.legend(title="Year"); plt.grid(alpha=.3)
plt.show()
```

## Graph 5: How chlorophyll drops off with distance from shore

```python
bins = np.arange(0, 610, 20)
season = {"Spring (Mar-May)": [3, 4, 5], "Summer (Jun-Aug)": [6, 7, 8],
          "Fall (Sep-Nov)": [9, 10, 11], "Winter (Dec-Feb)": [12, 1, 2]}
plt.figure(figsize=(9, 4.5))
for name, months in season.items():
    sub = chl[np.isin(dates.month, months)]
    prof = [logmean(sub[:, (dist_km >= b) & (dist_km < b + 20)]) for b in bins[:-1]]
    plt.plot(bins[:-1] + 10, prof, label=name)
plt.yscale("log"); plt.xlabel("Distance from shore (km)"); plt.ylabel("Chlorophyll-a (mg/m³)")
plt.title("Chlorophyll vs. distance offshore, by season"); plt.legend(); plt.grid(alpha=.3, which="both")
plt.show()
```

## Graph 6: Is colder water greener? (SST vs. chlorophyll, every pixel-month)

```python
x, y = sst.ravel(), chl.ravel()
m = np.isfinite(x) & np.isfinite(y) & (y > 0)
plt.figure(figsize=(7.5, 5.5))
plt.hexbin(x[m], y[m], yscale="log", gridsize=50, bins="log", cmap="magma_r", mincnt=1)
plt.colorbar(label="Pixel count (log)")
r = np.corrcoef(x[m], np.log10(y[m]))[0, 1]
plt.xlabel("Sea surface temperature (°C)"); plt.ylabel("Chlorophyll-a (mg/m³)")
plt.title(f"SST vs. chlorophyll, {YEARS[0]}-{YEARS[-1]} (r with log chl = {r:.2f})")
plt.show()
```

## Bonus graph: Coastal chlorophyll by latitude over time (great candidate for the D3 version)

```python
hov = np.array([[logmean(chl[t, r][near[r]]) if near[r].any() else np.nan for r in range(H)]
                for t in range(len(dates))]).T
plt.figure(figsize=(12, 5))
plt.imshow(hov, aspect="auto", cmap="viridis", norm=CHL_NORM,
           extent=[0, len(dates), REGION[0], REGION[2]])
ticks = [k for k, d in enumerate(dates) if d.month == 1]
plt.xticks(ticks, [dates[k].year for k in ticks])
plt.colorbar(label="Chlorophyll-a (mg/m³)"); plt.ylabel("Latitude")
plt.title(f"Coastal (< {NEARSHORE_KM} km) chlorophyll by latitude and month")
plt.show()
```

## Save aggregated data for the D3 version

```python
pd.DataFrame({"date": dates.strftime("%Y-%m"),
              "chl_coastal": chl_near, "chl_offshore": chl_off,
              "sst_coastal": [np.nanmean(s[near]) for s in sst],
              "sst_offshore": [np.nanmean(s[off]) for s in sst]}).to_csv("ocean_monthly.csv", index=False)
pd.DataFrame(hov, index=np.round(lats, 3), columns=dates.strftime("%Y-%m")).to_csv("coastal_chl_by_lat.csv")
print("saved ocean_monthly.csv, coastal_chl_by_lat.csv")
```
